# EJERCICIO 2
## Estadística descriptiva con NumPy

**Objetivo:** Calcular medidas estadísticas sobre la edad y la calificación de admisión.

**Actividades:**
- Calcular el promedio.
- Calcular la mediana.
- Identificar mínimo y máximo.
- Calcular desviación estándar.
- Contar estudiantes mayores de 25 y menores de 20 años.

```python
# Convertir datos en arreglos NumPy
edades = df['Age at enrollment'].to_numpy()
notas = df['Admission grade'].to_numpy()

# Estadísticas de edad
print("ESTADÍSTICAS DE EDAD")
print("Promedio:", round(np.mean(edades), 2))
print("Mediana:", np.median(edades))
print("Edad mínima:", np.min(edades))
print("Edad máxima:", np.max(edades))
print("Desviación estándar:", round(np.std(edades), 2))

# Contar estudiantes
print("Mayores de 25:", np.sum(edades > 25))
print("Menores de 20:", np.sum(edades < 20))

# Estadísticas de admisión
print("\nNOTAS DE ADMISIÓN")
print("Promedio:", round(np.mean(notas), 2))
print("Desviación estándar:", round(np.std(notas), 2))
print("Mínimo:", np.min(notas))
print("Máximo:", np.max(notas))
```

# EJERCICIO 3
## Distribución de los resultados académicos

**Objetivo:** Contar y calcular los porcentajes de los resultados registrados en Target.

**Actividades:**
- Contar Graduate, Dropout y Enrolled.
- Calcular porcentajes.
- Obtener la edad promedio de cada grupo.
- Identificar el grupo con mayor edad promedio.

```python
# Contar resultados
conteo = df['Target'].value_counts()

print("Cantidad de estudiantes:")
print(conteo)

# Calcular porcentajes
total = len(df)
porcentajes = conteo / total * 100

resumen = pd.DataFrame({
    'Cantidad': conteo,
    'Porcentaje': porcentajes
})

display(resumen.round(2))

# Edad promedio por resultado
edad = df.groupby('Target')['Age at enrollment'].mean()

print("\nEdad promedio:")
print(edad.round(2))

# Grupo con mayor edad
print("Mayor edad promedio:", edad.idxmax())
```

# EJERCICIO 4
## Comparación del rendimiento académico por semestre

**Objetivo:** Comparar las calificaciones y asignaturas aprobadas durante el primer y segundo semestre.

**Actividades:**
- Calcular los promedios de ambos semestres.
- Comparar las calificaciones por Target.
- Comparar asignaturas aprobadas.
- Calcular la diferencia de notas.
- Contar estudiantes que mejoraron, disminuyeron o mantuvieron su calificación.

```python
# Columnas de calificaciones
nota1 = 'Curricular units 1st sem (grade)'
nota2 = 'Curricular units 2nd sem (grade)'

aprobadas1 = 'Curricular units 1st sem (approved)'
aprobadas2 = 'Curricular units 2nd sem (approved)'

# Convertir a NumPy
semestre1 = df[nota1].to_numpy()
semestre2 = df[nota2].to_numpy()

# Promedios generales
print("Promedio primer semestre:",
      round(np.mean(semestre1), 2))

print("Promedio segundo semestre:",
      round(np.mean(semestre2), 2))

# Comparación por resultado
comparacion = df.groupby('Target')[
    [nota1, nota2, aprobadas1, aprobadas2]
].mean()

display(comparacion.round(2))

# Diferencia de notas
df['Diferencia_notas'] = df[nota2] - df[nota1]

diferencias = df['Diferencia_notas'].to_numpy()

print("Mejoraron:", np.sum(diferencias > 0))
print("Disminuyeron:", np.sum(diferencias < 0))
print("Mantuvieron:", np.sum(diferencias == 0))
```

# EJERCICIO 5
## Estado de matrícula y becas

**Objetivo:** Analizar el estado de pago de matrícula y la condición de becario frente al resultado académico.

**Actividades:**
- Contar matrículas al día y pendientes.
- Calcular porcentajes.
- Crear tablas cruzadas.
- Contar estudiantes con beca y sin beca.

```python
# Obtener datos
matricula = df['Tuition fees up to date'].to_numpy()
beca = df['Scholarship holder'].to_numpy()

# Estado de matrícula
al_dia = np.sum(matricula == 1)
pendiente = np.sum(matricula == 0)

print("Matrícula al día:", al_dia)
print("Matrícula pendiente:", pendiente)

print("Porcentaje al día:",
      round(al_dia / len(df) * 100, 2), "%")

print("Porcentaje pendiente:",
      round(pendiente / len(df) * 100, 2), "%")

# Estudiantes con y sin beca
print("Con beca:", np.sum(beca == 1))
print("Sin beca:", np.sum(beca == 0))

# Tablas cruzadas
print("\nMatrícula y resultado:")
display(pd.crosstab(
    df['Tuition fees up to date'],
    df['Target']
))

print("\nBecas y resultado:")
display(pd.crosstab(
    df['Scholarship holder'],
    df['Target']
))
```

# EJERCICIO 7
## Miniinforme: factores asociados al resultado académico

**Objetivo:** Integrar Pandas, NumPy y Matplotlib para realizar un análisis exploratorio de los datos.

**Actividades:**
- Seleccionar variables académicas.
- Calcular la matriz de correlación.
- Calcular la correlación entre los semestres.
- Crear un gráfico de correlación.
- Redactar cinco hallazgos y dos recomendaciones.

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Age at enrollment'],
         bins=15, edgecolor='black')

plt.title('Distribución de edades')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')

plt.grid(axis='y')
plt.show()
```

```python
promedios = df.groupby('Target')[
    ['Curricular units 1st sem (grade)',
     'Curricular units 2nd sem (grade)']
].mean()

promedios.plot(kind='bar', figsize=(9, 5))

plt.title('Promedio de notas por semestre')
plt.xlabel('Resultado académico')
plt.ylabel('Promedio')
plt.xticks(rotation=0)

plt.legend(['Primer semestre', 'Segundo semestre'])
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Curricular units 1st sem (grade)'],
    df['Curricular units 2nd sem (grade)'],
    alpha=0.35
)

plt.title('Comparación de calificaciones')
plt.xlabel('Nota primer semestre')
plt.ylabel('Nota segundo semestre')

plt.grid()
plt.show()
```

```python
# Seleccionar variables
variables = [
    'Age at enrollment',
    'Admission grade',
    'Curricular units 1st sem (grade)',
    'Curricular units 1st sem (approved)',
    'Curricular units 2nd sem (grade)',
    'Curricular units 2nd sem (approved)'
]

# Matriz de correlación
matriz_corr = df[variables].corr()

display(matriz_corr.round(2))

# Correlación con NumPy
nota1 = df['Curricular units 1st sem (grade)'].to_numpy()
nota2 = df['Curricular units 2nd sem (grade)'].to_numpy()

corr_numpy = np.corrcoef(nota1, nota2)[0, 1]

print("Correlación entre los semestres:",
      round(corr_numpy, 4))

# Gráfico de correlación
plt.figure(figsize=(9, 7))

plt.imshow(matriz_corr, cmap='coolwarm',
           vmin=-1, vmax=1)

plt.colorbar(label='Correlación')

plt.xticks(range(len(variables)), variables,
           rotation=75, ha='right')

plt.yticks(range(len(variables)), variables)

for i in range(len(variables)):
    for j in range(len(variables)):
        plt.text(
            j, i,
            f'{matriz_corr.iloc[i, j]:.2f}',
            ha='center',
            va='center',
            fontsize=8
        )

plt.title('Matriz de correlación académica')
plt.tight_layout()
plt.show()

# Resumen final
resumen_final = df.groupby('Target').agg(
    estudiantes=('Target', 'size'),
    edad_promedio=('Age at enrollment', 'mean'),
    nota_semestre_1=('Curricular units 1st sem (grade)', 'mean'),
    nota_semestre_2=('Curricular units 2nd sem (grade)', 'mean'),
    aprobadas_semestre_1=('Curricular units 1st sem (approved)', 'mean'),
    aprobadas_semestre_2=('Curricular units 2nd sem (approved)', 'mean')
)

display(resumen_final.round(2))
```
